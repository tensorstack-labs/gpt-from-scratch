```python
import torch
import torch.nn as nn

print(f"✔ PyTorch Version: {torch.__version__}")
```

### 1. Discrete Text to Continuous Vector Projection

```python
sample_words = ["Neural", "networks", "learn", "patterns"]
vocab = {w: i for i, w in enumerate(sample_words)}

token_ids = torch.tensor([vocab[w] for w in sample_words])
print("Token IDs:", token_ids.tolist())

# 768-dimensional Continuous Embedding Table
d_model = 768
embed = nn.Embedding(len(vocab), d_model)
vectors = embed(token_ids)
print(f"Continuous Vectors Shape: {vectors.shape}  # [Num_Tokens, d_model]")
```
