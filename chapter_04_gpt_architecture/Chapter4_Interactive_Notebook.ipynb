Welcome to Chapter 4! In this notebook, we code and instantiate the complete 124-Million parameter GPT-2 architecture and run autoregressive text generation.

```python
import torch
from gpt_architecture import GPTModel, GPT_CONFIG_124M, generate_text_simple

torch.manual_seed(123)
model = GPTModel(GPT_CONFIG_124M)
total_params = sum(p.numel() for p in model.parameters())
print(f'Total Trainable Parameters: {total_params:,}')

# Test Forward Pass
sample_tokens = torch.randint(0, 50257, (1, 8))
logits = model(sample_tokens)
print('Logits shape:', logits.shape)

# Generate 10 New Tokens
gen_out = generate_text_simple(model, sample_tokens, max_new_tokens=10, context_size=1024)
print('Generated sequence:', gen_out)
```
